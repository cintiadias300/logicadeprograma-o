```python
import math
print("=== ÁREA E PERÍMETRO DO CÍRCULO ===")

raio = float(input("Digite o raio do círculo: "))

area = math.pi * raio ** 2
perimetro = 2 * math.pi * raio

print(f"Área do círculo: {area:.4f}")
print(f"Perímetro do círculo: {perimetro:.4f}")
print("\n=== EQUAÇÃO DO 2º GRAU ===")

a = float(input("Digite o valor de a (diferente de zero): "))
b = float(input("Digite o valor de b: "))
c = float(input("Digite o valor de c: "))

if a == 0:
    print("Erro: o valor de a deve ser diferente de zero.")
else:
    delta = b ** 2 - 4 * a * c

    if delta > 0:
        raiz_delta = math.sqrt(delta)

        x1 = (-b + raiz_delta) / (2 * a)
        x2 = (-b - raiz_delta) / (2 * a)

        print(f"Delta: {delta:.4f}")
        print(f"Primeira raiz (x1): {x1:.4f}")
        print(f"Segunda raiz (x2): {x2:.4f}")

    elif delta == 0:
        print("Delta igual a zero: existe uma raiz real.")
    else:
        print("Delta negativo: não existem raízes reais.")

print("\n=== ERRO ABSOLUTO DE √2 ===")

raiz_2 = math.sqrt(2)
aproximacao = 1.4142

erro_absoluto = abs(raiz_2 - aproximacao)

print(f"Valor de √2: {raiz_2:.10f}")
print(f"Aproximação: {aproximacao}")
print(f"Erro absoluto: {erro_absoluto:.10f}")
```
